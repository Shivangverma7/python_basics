### __Loops__
- for loop
- while loop

```python
list = [3,4,5, 6,7,7,45,786,]
for i in list:
    print(i)
```

```python
list = [3,4,5, 6,7,7,45,786,]
for i in list:
    print(i, end='*')
```

```python
name = 'Alice Roy'
for i in name:
    print(i, end='')
```

```python
for i in range(21):
    print(i)
```

```python
for i in range(1,21):
    print(i)
```

```python
for i in range(1,21,2):
    print(i)
```

- print all the even numbers till 20 without using conditions statements

```python
for i in range(5):
    print(f" {i+1} shivang" )
```

```python
for i in range(5):
    print("shivang" )
```

```python
for i in range(21):
    if i%2 != 0:
        print(i)
```

```python
for i in range(0,21,2):
    print(i)
```

```python
for i in range(20, 0, -1):
    print(f" {i+1} shivang" )
```

```python
# While loop
x = 0
while x<21:
    print(x)
    x+=1
```

```python
y = 0
while y < 5:
    print("Shivang")
    y +=1
```

```python
z = 0
while z <20:
    if z %2 != 0:
        print(z)
    z +=1
```
