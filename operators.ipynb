### __Operators in Python__
Operators are symbols that perform operations on values or variables (called operands).

### 1️⃣ Arithmetic Operators
Used for mathematical calculations

```python
a = 30
b = 20
print(a+b)
print(a-b)
print(a*b)
print(a/b)
print(a//b)
print(a%b)
print(a**b)
```

### 2️⃣ Assignment Operators
Used to assign values.

```python
a = 5  # assigned
print(a)
a += 3 # sameas a = a + 3
print(a)
a -= 3 # sameas a = a - 3
print(a)
a *= 3 # sameas a = a * 3
print(a)
a /= 3 # sameas a = a / 3
print(a)
```

### 3️⃣ Comparison (Relational) Operators
Used to compare two values.

```python
a = 10
b = 5
print(a == b) # Equal to
print(a != b) # Not Equal to
print(a > b) # Greater than
print(a < b) # Less than
print(a >= b) # Greater than or equal to
print(a <= b) # less than or equal to
```

### 4️⃣ Logical Operators
Used to combine conditions.
> and , or , not

```python
x = 9 
y = 7
z = 11
print(x>y and x>z) # True if both conditions are true
print(x>y or x>z)  # True if at least one is true
print(not(x>y and x>z)) # Reverses the result
```

```python
x = 10 
y= 7
z = 11
not(x>y and x>z)
```

### 5️⃣ Membership Operator 
Check if a value exists in a sequence.
>in / not in

```python
list = [1,2, 3, 4, 6, 7,]

print(6 in list)
print(5 in list)
print("s" in "shivang")
print("s" not in "shivang")
print("a" in "apple")
print("a" not in "apple")   
```

#### 6️⃣ Identity Operators
Compare memory locations.
>is / is not

```python
a = 10
b = 10
print(a is b)
print(a is not b)
```
