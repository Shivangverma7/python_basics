### __Conditional Statements__
___

`Types of Conditions and their statements`
- Single Condition
    - if + else
- Multiple Conditions
    - if
    - else
    - nested if


```python
marks = 85
if marks >= 90:
    print('A+')
if marks >= 75:
    print('A')
if marks >= 60:
    print('B')
if marks < 60:
    print('fail')
```

```python
marks = 85
if marks >= 90:
    print('A+')
elif marks >= 75:
    print('A')
elif marks >= 60:
    print('B')
elif marks < 60:
    print('fail')
else:
    print('Invalid marks')
```
